2.Validació de dades i rangs







```python
# funció per a validar la llista dels candidats procesats

def validar_dades(candidats_procesats):

    candidats_filtrats = [] # creem nova llista per als candidats correctes
    i = 0

    while i < len(candidats_procesats): # fem un bucle per a que agafi tota la llista
        candidat = candidats_procesats[i] # indiquem la posició primera de la llista

        # validem tots els elements del diccionari de cada candidat
        if (candidat["nom"] != "" and
            str(candidat["telefon"]) != "" and
            candidat["anys_experiencia"] >= 0 and
            0 <= candidat["coneixements_catala"] <= 3 and
            0 <= candidat["domini_excel"] <= 3 and
            0 <= candidat["titulacions"] <= 3):

            candidats_filtrats.append(candidat) # afegim els correctes a la llista
            print(f"correcte ({len(candidats_filtrats)}/10): {candidat['nom']}")

            if len(candidats_filtrats) == 10:
                break # parem la validació quan arribem als 10
        else:
            print(f"incorrecte: {candidat['nom']}")
             # afegim incorrecte per als que deixem fora de la llista

        i += 1  # per a que sumi 1 a i cada vegada


    return candidats_filtrats
```

```python
# comprovem el resultat de tots els candidats
resultat = validar_dades(candidats_procesats)
print(resultat)
```

3.Càlcul de punts







```python
# definim funció per obtenir el càlcul de punts per candidat
def calcular_punts(candidat):

  punts_totals = (
      candidat["anys_experiencia"] * 1.5 +
      candidat["coneixements_catala"] +
      candidat["domini_excel"] +
      candidat["titulacions"]
  )
  candidat["puntuació"] = punts_totals # afegim puntuació al diccionari del candidat

  return punts_totals
```

```python
# definim la funció per a classificar al candidat amb els paràmetres establerts i l'afegim al seu diccionari.

def classificar_usuari(candidat):

    punts_totals = candidat["puntuació"] #li diem que agafi la puntuació

    if punts_totals >= 12:
        classificacio = "Prioritat alta"
    elif punts_totals >= 8:
        classificacio = "Candidat adequat"
    else:
        classificacio = "No prioritari"

    candidat["classificacio"] = classificacio # l'afegim al diccionari

    return classificacio
```

```python
# definim la funció principal que orquestri totes les altres

def main_seleccio(llista_test):
     # cridem a totes les funcions
    candidats_procesats = obtenir_dades(llista_test)
    candidats_procesats = validar_dades(candidats_procesats)

    for candidat in candidats_procesats: # bucle per als càlculs de cada candidat dins la llista de candidats proces
        calcular_punts(candidat)
        classificar_usuari(candidat)
        mostrar_resultat(candidat)

    df = pd.DataFrame(candidats_procesats)

    df_final = df.sort_values(by=['puntuació', 'classificacio'], ascending=[False, True]) # endreçem els valors de puntuació i classificació de major a menor

    return df_final
```

```python
#diccionari on tenim l'edat i el factor segons el grup d'edat dels clients

edat_factor = {
"18-25": 1.20,
 "26-40": 1.00,
 "41-60": 0.85,
 "60+": 0.70
}
```

```python
# definim la funció per calcular les dades dels clients i poder-los porcesar

def calcular_dades_clients(dades_clients):
    for client in dades_clients: # bucle que recorre tots els clients
        codi = client["codi_comportament"]

        # obtenim segment i factor i l'afegim al diccionari com a nous elements
        client["factor_codi"] = codi_factor[codi]["factor"]
        client["segment"] = codi_factor[codi]["segment"]
        client["missatge"] = codi_factor[codi]["missatge"]

        # obtenim factor edat i l'afegim al diccionari
        grup = client["grup_edat"]
        client["factor_edat"] = edat_factor[grup]

        # calculem risc final i l'afegim al diccionari
        client["risc_final"] = round(client["factor_codi"] * client["factor_edat"], 2)

    return dades_clients
```

🎯 Objectius
L'objectiu és transformar aquestes dades en informació útil per als metges. A partir dels registres en brut, has de:

» Convertir cada string en una estructura neta (diccionari o similar).

» Separar i normalitzar la llista de

símptomes:

minúscules

eliminar espais

dividir correctament per comes

» Assignar un risc a cada símptoma

utilitzant el diccionari nivells_risc.

```python
#llista amb informació de pacients

pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
```

```python
#diccionari de nivell de riscs

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

**Definició de l'Algorisme**

1.Crearem una funció per netejar dades def netejar_dades. Aquesta funció rebrà la llista de pacients i farà el següent:



*   Normalitzar dades: convertir en minúscula les lletres, eliminar espais, treure les barres separadores i separar les paraules correctament amb una coma.

Aquesta funció retornarà un diccionari dels pacients normalitzat per a poder treballar les dades correctament.

2.Crear una funció per assignar un risc a cada símptoma utilitzant el diccionari de pacients i nivells_risc.Rep un pacient i retorna un pacient amb el seu risc al diccionari.

3.Funció per calcular el risc total de cada pacient sumant els valors.Aquesta funció rep el diccionari de pacients i fa la suma dels valors del nivell de risc.Retorna la suma al camp "risc_total" i l'afegeix al diccionari del pacient.

4.Endreçar el pacient de majora a menor risc i convertir-ho a dataframe.

Mostrar resultats en format taula en un dataframe que compleixi els següents requisits:

<<nom

<<símptomes normalitzats

<<risc_total

<<apartat final amb els pacients ordenats per prioritat (més risc → primer).


1. Creeem la funció per netejar les dades dels pacients.





```python
# definim la funció per netejar les dades dels pacients
def netejar_dades (pacients_llista_neta):

  pacients_procesats = [] #nova llista per als pacients procesats

  # bucle per netejar les dades
  for pacient_str in pacients_llista_neta:

    parts = pacient_str.split('|')
    nom = parts[0].strip()
    edat = int(parts[1].strip())

    simptomes_original_str = parts[2].strip()
    simptomes_lower_str = simptomes_original_str.lower()

    simptomes_list_format = [s.strip() for s in simptomes_lower_str.split(',')]
    simptomes = [s.replace(' ', '_') for s in simptomes_list_format]

    pacient_dict = {
        "nom": nom,
        "edat": edat,
        "simptomes": simptomes,
    }
    pacients_procesats.append(pacient_dict)

  return pacients_procesats
```

2. Funció asignar sic

```python
# Definim la funció per assignar el risc a cada símptoma d'un pacient
def assignar_risc_simptomes(un_pacient, llistat_riscos):
    riscos_simptomes = []
    for simptoma in un_pacient['simptomes']:
        # Obtenim el risc o 0 si el símptoma no és a la llista
        riscos_simptomes.append(llistat_riscos.get(simptoma, 0))

    un_pacient['riscos_simptomes'] = riscos_simptomes
    return un_pacient

```

3. Calcular risc dels pacients

```python
def calcular_risc_total(un_pacient):
    # Sumem els números de la llista 'riscos_simptomes' d'aquest pacient concret
    un_pacient['risc_total'] = sum(un_pacient['riscos_simptomes'])

    # Retornem el pacient sencer modificat
    return un_pacient
```

4. Funció ordenar pacients

```python
def ordenar_pacients(llista_pacients):
    df_pacients = pd.DataFrame(llista_pacients)
    df_ordenat = df_pacients.sort_values(by='risc_total', ascending=False)
    return df_ordenat
```

5.Crear la funció principal que orquestri totes les altres i mostrar dataframe

```python

# definim funció principal
def main_pacients(pacients, nivells_risc):
    # Pas 1: Netegem les dades (retorna llista de diccionaris)
    pacients_procesats = netejar_dades(pacients)

    # assignem el risc i calculem el total modificant els diccionaris
    for pacient in pacients_procesats:
        assignar_risc_simptomes(pacient, nivells_risc)
        calcular_risc_total(pacient)

    # cridem a la funció que converteix a DataFrame i ordena
    pacients_ordenats_df = ordenar_pacients(pacients_procesats)

    return pacients_ordenats_df


# Cridem la funció principal per obtenir directament el DataFrame ordenat
df_pacients = main_pacients(pacients, nivells_risc)

# Triem les columnes que volem tractar
df_taula = df_pacients[['nom', 'edat', 'simptomes', 'risc_total']].copy()
df_taula.columns = ['nom', 'edat', 'símptomes normalitzats', 'risc_total']



# Ajuntem les llistes amb comes i canviem els guions de forma segura
df_taula['símptomes normalitzats'] = df_taula['símptomes normalitzats'].apply(
    lambda x: ', '.join(x).replace('_', ' ')
)

# Funció per afegir les boletes dels colors del semàfor
def obtenir_bolita_risc(risc):
    if risc <= 3:
        return "🟢"
    elif risc <= 7:
        return "🟡"
    else:
        return "🔴"

# Creem la nova columna nivell de risc
df_taula['nivell de risc'] = df_taula['risc_total'].apply(obtenir_bolita_risc)


# afegim semàfor de colors per assignar el risc
print("PACIENTS ORDENATS PER PRIORITAT DE RISC")
print("Llegenda: 🔴 Alt (Risc > 7)  |  🟡 Mig (Risc 4-7)  |  🟢 Baix (Risc 0-3)")
print("-" * 75)

# Triem les columnes finals i les mostrem
columnes_finals = ['nom', 'edat', 'símptomes normalitzats', 'risc_total', 'nivell de risc']
display(df_taula[columnes_finals])
```

**Definició Algorisme**

1.Hem de definir una funció per obtenir les dades que necessitem del diccionari de clients (edat i codi). Aquesta funció rep diccionari de client i nivells de seguro, afegeix la categoria, el preu base i la probabilitat de frau a la informació del client com a columnes noves.

2.Aquesta funció rep el diccionari del client (que ja està actualitzat amb el preu base). Aquesta funció aplica la fórmula matemàtica preu_base×(1+edat/100) per calcular l'import final de la pòlissa, afegeix aquest valor directament al client com a 'preu_final' (arrodonint a dos decimals) i el retorna.

3.Definim una funció principal que orquestri les altres.

4.Mostrar el dataframe ordenat per probabilitat de frau i preu.
  
Mostrar anàlisis i conclusions basant-nos en les següents preguntes:

Quin client té més probabilitat de frau

Quin té el preu final més alt

Qui seria prioritari de revisar manualment


1.Cream funció per obtenir dades

```python
# funció per procesar clients

def procesar_client(un_client, nivells_seg):

    # definim les variables
    nom = un_client["nom"]
    edat = un_client["edat"]
    codi = un_client["codi"]

    # Busquem al diccionari de nivells de segur local
    info_seguros = nivells_seg[codi]

    categoria = info_seguros["categoria"]
    preu_base = info_seguros["preu_base"]
    frau = info_seguros["frau"]

    client_procesat = {    #creem un diccionari
        "nom": nom,
        "edat": edat,
        "categoria": categoria,
        "preu_base": preu_base,
        "probabilitat_frau": frau
    }
    return client_procesat
```

2. Càlcul del preu final

```python
# funció per calcular el preu final

def calcular_preu_final(un_client_processat):
    preu_base = un_client_processat["preu_base"]
    edat = un_client_processat["edat"]

    # Apliquem la fórmula matemàtica i arrodonim a 2 decimals
    preu_final = round(preu_base * (1 + edat / 100), 2)

    un_client_processat["preu_final"] = preu_final
    return un_client_processat
```

3. definició de funció principal i mostra de dataframe

```python
# definim funció principal
def main_seguros(clients, nivells_seguro):
    clients_finals = []

    for client in clients:
        # processem les dades bàsiques i de frau
        client_processat = procesar_client(client, nivells_seguro)

        # calculem el preu final i l'afegim al client
        client_amb_preu = calcular_preu_final(client_processat)

        # guardem resultats a llista nove
        clients_finals.append(client_amb_preu)

    return clients_finals

```

4.Mostrar dataframe final i anàlisis

```python


def mostrar_dataframe(llista_clients, diccionari_seguros):
    # processem tots els clients amb la funció principal
    dades_finals_clients = main_seguros(llista_clients, diccionari_seguros)

    # creem datafram amb les dades calculades i procesades
    df_clients = pd.DataFrame(dades_finals_clients)

    # ordenem els clients per probabilitat de frau i preu final descendent
    df_ordenat = df_clients.sort_values(
        by=['probabilitat_frau', 'preu_final'],
        ascending=[False, False]
    )

    # mostrem el dataframe amb les columnes que demana l'enunciat
    print("DATAFRAME DE CLIENTS ORDENATS PER FRAU I PREU")
    print("-" * 75)

    columnes_finals = ['nom', 'edat', 'categoria', 'preu_base', 'preu_final', 'probabilitat_frau']
    display(df_ordenat[columnes_finals])

    print("\n" + "="*75 + "\n")

    # anàlisi i resposta a les preguntes
    client_max_frau = df_ordenat.iloc[0]  # el primer de la llista ordenada per frau és el de més risc
    client_max_preu = df_clients.loc[df_clients['preu_final'].idxmax()]  # el preu de pòlissa més alt

    print("APARTAT D'ANÀLISI I CONCLUSIONS")
    print("-" * 75)

    # Mostrem la probabilitat de frau (si és un decimal com 0.8, ho multipliquem per 100)
    frau_pct = client_max_frau['probabilitat_frau']
    if frau_pct <= 1.0:
        frau_display = f"{int(frau_pct * 100)}%"
    else:
        frau_display = f"{int(frau_pct)}%"

    print(f"» Client amb més probabilitat de frau: {client_max_frau['nom']} ({frau_display})")
    print(f"» Client amb el preu final més alt:   {client_max_preu['nom']} ({client_max_preu['preu_final']} €)")
    print(f"» Prioritat de revisió manual:         {client_max_frau['nom']} (Risc crític i alta probabilitat de frau)")


mostrar_dataframe(clients, nivells_seguro)
```
