### Codificació

Detallem pas a pas els problemes resolts:

- **Problema detectat:** diversos noms de columna i valors de text contenien caràcters estranys (`MuÃ±oz`, `CÃ rrec`, `â‚¬`...), típic d'un text guardat en UTF-8 però interpretat incorrectament amb una altra codificació en llegir-lo (fenomen conegut com a *mojibake*).
- **Solució aplicada:** es reverteix l'error d'interpretació amb `.encode('cp1252').decode('utf-8')` — es recuperen els bytes originals (assumint que es van llegir malament com a Windows-1252/cp1252) i es tornen a interpretar correctament com a UTF-8. *(S'ha fet servir cp1252 en comptes de Latin-1 perquè alguns caràcters, com el símbol de l'euro mal codificat, només existeixen en el rang de cp1252, no en Latin-1.)*
- **Per què la funció té `isinstance` i `try/except`:**
  - `isinstance(valor, str)` evita aplicar la correcció a valors que no són text (booleans, `NaN`), que no tenen el mètode `.encode()`.
  - El `try/except UnicodeError` evita que el procés s'aturi si algun text està tan malmès que no es pot corregir (per exemple, truncat) — en aquests casos, es deixa el valor tal com està temporalment, sense frenar la resta de la neteja.
- **Tractament dels casos irreparables:** després d'aplicar la correcció, es comprova si encara queda algun rastre de mojibake (patró `Ã`). Si és així, es considera informació irrecuperable i es marca com a valor nul (`NaN`), en comptes de deixar text corromput al dataset.
- **Aplicació:** la funció s'aplica tant als valors de les columnes de text (fent servir `.apply()` en un bucle sobre les columnes `str`) com als propis noms de les columnes (`df.columns`).

```python
# Funció per corregir problemes de codificació (mojibake UTF-8 llegit com Latin-1).
# Si el valor no és text, es retorna tal qual. Si el text és irrecuperable
# (truncat o amb caràcters que no existeixen en Latin-1), també es retorna tal qual,
# gràcies al try/except, en comptes d'aturar tot el procés.

def arreglar_encoding(valor):
    if isinstance(valor, str):
        try:
            return valor.encode('cp1252').decode('utf-8')
        except UnicodeError:
            return valor
    else:
        return valor

# Apliquem la correcció a totes les columnes de tipus text (valors de les cel·les)
columnas_texto = df.select_dtypes(include=['str']).columns
for col in columnas_texto:
    df[col] = df[col].apply(arreglar_encoding)

# Apliquem la mateixa correcció als noms de les columnes

df.columns = [arreglar_encoding(col) for col in df.columns]

```

```python
# Netejem qualsevol resta de mojibike que no es pugui reparar (marquem com a nul)

for col in df.select_dtypes(include=['str']).columns:
    df[col] = df[col].apply(lambda x: np.nan if isinstance(x, str) and 'Ã' in x else x)
```

```python
# Comprovació final de les columnes

for col in df.select_dtypes(include=['str']).columns:
    restantes = df[col].astype(str).str.contains('Ã', na=False).sum()
    if restantes > 0:
        print(col, '->', restantes)
```

### DNI's duplicats

- **Problema detectat:** s'han trobat 16 files implicades en DNIs repetits, agrupades en dos patrons diferents:
  - Casos on el DNI es repeteix per a la **mateixa persona** (coincideixen Nom i Any de Naixement), probablement per registres desactualitzats o introduïts més d'un cop.
  - Casos on el mateix DNI apareix associat a **persones clarament diferents** (noms, ciutats i dates de naixement incompatibles) — un error greu, ja que un DNI no pot pertànyer a més d'una persona.
- **Criteri aplicat:**
  - Quan es tracta de la mateixa persona, es conserva únicament la fila amb menys valors nuls (la més completa/fiable), i s'elimina la resta.
  - Quan es tracta de persones diferents amb el mateix DNI, es considera que cap de les files és fiable (no es pot determinar quina és la correcta), i s'eliminen totes les files del grup.
- **Resultat:** s'han eliminat **13 files** per DNI duplicat no fiable.

```python
df['DNI'].duplicated().sum()
```

```python
df[df['DNI'].notna() & df['DNI'].duplicated(keep=False)].sort_values('DNI')
```

```python
# Guardem el nombre de files abans de resoldre els DNI duplicats
files_abans_dni = df.shape[0]

# Funció per decidir, dins d'un grup amb el mateix DNI, si són la mateixa persona
# (coincideixen Nom i Any de Naixement) o persones diferents.
def resoldre_grup(grup):
    mateixa_persona = (grup['Nom'].nunique() == 1) and (grup['Any de Naixement'].nunique() == 1)
    if mateixa_persona:
        # Ens quedem amb la fila amb menys valors nuls (més informació completa)
        idx_millor = grup.notna().sum(axis=1).idxmax()
        fila = grup.loc[[idx_millor]].copy()
        fila['DNI'] = grup.name  # reconstruïm la columna DNI (Pandas 3.x l'exclou de "grup")
        return fila
    else:
        # Persones diferents amb el mateix DNI: DNI no fiable, eliminem tot el grup
        buida = grup.iloc[0:0].copy()
        buida['DNI'] = []
        return buida

sense_duplicats = df[~df['DNI'].duplicated(keep=False)]
amb_duplicats = df[df['DNI'].duplicated(keep=False)]

resolts = amb_duplicats.groupby('DNI', group_keys=False).apply(resoldre_grup)
df = pd.concat([sense_duplicats, resolts]).sort_index()

files_eliminades_dni = files_abans_dni - df.shape[0]
print(f"S'han eliminat {files_eliminades_dni} files per DNI duplicat no fiable")
```

### Coherència Te_cotxe / Km_anuals / Consum_mitjà

- **Problema detectat:** incoherències entre columnes relacionades amb el cotxe — persones amb `Km_anuals` o `Consum_mitjà` registrat però marcades com a `Te_cotxe = False` o buit, i casos amb `Consum_mitjà` però sense `Km_anuals`.
- **Solució aplicada:** s'ha imputat `Km_anuals` amb la mitjana general en els casos on constava `Consum_mitjà` però no `Km_anuals` (2 files), i s'ha corregit `Te_cotxe` a `True` quan hi havia quilometratge registrat (4 files), ja que la presència d'aquest valor implica necessàriament tenir cotxe.

```python
# Comprovem la coherència entre Te_cotxe, Km_anuals i Consum_mitja_L_100km:
# si algú té quilometratge anual o consum mitjà, hauria de tenir cotxe (Te_cotxe = True).
te_km = df['Km_anuals'].notna()
te_consum = df['Consum_mitja_L_100km'].notna()

# Casos on diu que té cotxe però no té ni quilometratge ni consum registrat
print("Te_cotxe=True però sense Km ni Consum:", ((df['Te_cotxe'] == True) & ~te_km & ~te_consum).sum())

# Casos on NO té cotxe (o no consta) però sí té quilometratge registrat -> incoherent
print("Te_cotxe=False/NaN però amb Km:", ((df['Te_cotxe'] != True) & te_km).sum())

# Casos on hi ha consum mitjà però no quilometratge -> incoherència esmentada explícitament a l'enunciat
print("Té Consum però NO té Km:", (te_consum & ~te_km).sum())

# Casos on hi ha quilometratge però no consum -> possible dada incompleta a completar
print("Té Km però NO té Consum:", (te_km & ~te_consum).sum())
```

```python
# Per als casos on hi ha Consum però no Km, imputem el Km_anuals amb la mitjana
# general de la resta d'empleats que sí tenen aquest valor registrat.
mitjana_km = df['Km_anuals'].mean()

mask_cas3 = te_consum & ~te_km
df.loc[mask_cas3, 'Km_anuals'] = mitjana_km

print(f"S'ha imputat el Km_anuals (mitjana={mitjana_km:.0f}) a {mask_cas3.sum()} files")
```

```python
# Si algú té Km_anuals registrat, per definició ha de tenir cotxe.
# Corregim Te_cotxe a True en aquests casos (la incoherència estava en aquesta columna).
mask_cas2 = (df['Te_cotxe'] != True) & te_km
df.loc[mask_cas2, 'Te_cotxe'] = True

print(f"S'ha corregit Te_cotxe a True en {mask_cas2.sum()} files (tenien Km registrat)")
```

- **Resultat:** 241 files imputades a 0, i el dataset redueix la seva redundància sense perdre informació.

```python
# Comprovem coherència: si Fills=True, hauria d'haver-hi Nombre_fills >= 1
# Si No Fills=True (o Fills=False), Nombre_fills hauria de ser 0
incoherencia_1 = (df['Fills'] == True) & (df['Nombre_fills'].fillna(0) < 1)
incoherencia_2 = (df['No Fills'] == True) & (df['Nombre_fills'].fillna(0) > 0)

print("Fills=True però Nombre_fills < 1:", incoherencia_1.sum())
print("No Fills=True però Nombre_fills > 0:", incoherencia_2.sum())
```

```python
# Unifiquem la informació de fills en una única columna: Nombre_fills.
# Criteri:
# 1. Si Nombre_fills ja té un valor, el mantenim (és la dada més fiable i específica).
# 2. Si Nombre_fills és NaN però No Fills = True, sabem que no té fills -> 0.
# 3. Si Nombre_fills és NaN però Fills = True, sabem que té fills però no quants ->
#    s'imputa amb la mitjana de fills entre les persones que sí en tenen (>0).
# 4. Si no hi ha cap informació a cap de les tres columnes, es queda com a NaN.

mitjana_fills = df.loc[df['Nombre_fills'] > 0, 'Nombre_fills'].mean()

# Cas 2: No Fills = True i Nombre_fills buit -> 0
mask_no_fills = (df['No Fills'] == True) & df['Nombre_fills'].isna()
df.loc[mask_no_fills, 'Nombre_fills'] = 0

# Cas 3: Fills = True i Nombre_fills buit -> mitjana
mask_fills_sense_num = (df['Fills'] == True) & df['Nombre_fills'].isna()
df.loc[mask_fills_sense_num, 'Nombre_fills'] = mitjana_fills

print(f"Casos 'No Fills' sense número, imputats a 0: {mask_no_fills.sum()}")
print(f"Casos 'Fills' sense número, imputats amb mitjana ({mitjana_fills:.1f}): {mask_fills_sense_num.sum()}")

# Eliminem les columnes redundants Fills i No Fills, ja que Nombre_fills ja conté tota la informació
df = df.drop(columns=['Fills', 'No Fills'])

print(f"Columnes finals: {df.shape[1]}")
```

### Neteja i imputació de Salari mensual

- **Problema detectat:** la columna `Salari mensual` era de tipus text, amb múltiples formats inconsistents (separador de milers amb coma o punt, símbol € o paraula "euros", amb o sense espai), i alguns valors escrits directament en paraules (`mil vuit-cents`, `dos mil`, `tres mil`) o sense sentit (`n/d`).
- **Solució aplicada:** es converteix tota la columna a un valor numèric (`float`), traduint els valors en paraules al seu equivalent numèric, eliminant símbols de moneda i separadors de milers, i marcant com a nul els valors irrecuperables (`n/d`). Els salaris que falten s'imputen amb la mitjana salarial del mateix `Càrrec`, ja que persones amb el mateix lloc de treball solen tenir salaris similars.
- **Resultat:** la columna queda completament numèrica i sense valors nuls, llesta per a l'anàlisi.

```python
# Explorem els valors únics de Salari mensual abans de netejar-los,
# per detectar tots els formats diferents presents a la columna.
print("Tipus de dada actual:", df['Salari mensual'].dtype)
print("Nombre de valors únics:", df['Salari mensual'].nunique())
print(df['Salari mensual'].unique()[:40])

# Comprovem si hi ha valors que NO contenen cap dígit (escrits en paraules)
import re
valors_en_paraules = [v for v in df['Salari mensual'].dropna().unique() if not re.search(r'\d', str(v))]
print("\nValors escrits en paraules (sense cap dígit):", valors_en_paraules)
```

```python
# Netegem i convertim Salari mensual a un valor numèric (float).
# Passos:
# 1. Substituïm els valors escrits en paraules (català) pel seu equivalent numèric.
# 2. Marquem com a nul els valors sense sentit ("n/d").
# 3. Eliminem el símbol/paraula de moneda (€, euros) i els espais.
# 4. Eliminem els separadors de milers (comes i punts, mai representen decimals aquí).
# 5. Convertim el resultat a float.

mapeo_paraules = {
    'mil vuit-cents': '1800',
    'dos mil': '2000',
    'tres mil': '3000',
    'n/d': None,
}

def netejar_salari(valor):
    if pd.isna(valor):
        return None
    valor = str(valor).strip()
    if valor in mapeo_paraules:
        valor = mapeo_paraules[valor]
        if valor is None:
            return None
    valor = valor.replace('€', '').replace('euros', '').replace(',', '').replace('.', '').strip()
    try:
        return float(valor)
    except ValueError:
        return None

salaris_abans = df['Salari mensual'].copy()
df['Salari mensual'] = df['Salari mensual'].apply(netejar_salari)

print(df['Salari mensual'].describe())
```

```python
# Comprovem si hi ha algun Càrrec sense cap salari vàlid (per assegurar que la mitjana és calculable)
nuls_abans_salari = df['Salari mensual'].isna().sum()

mitjanes_per_carrec = df.groupby('Càrrec')['Salari mensual'].mean()
print(mitjanes_per_carrec)
```

```python
# Imputem els salaris faltants amb la mitjana salarial del mateix Càrrec,
# ja que persones amb el mateix lloc de treball solen tenir salaris similars.
# Arrodonim a 2 decimals perquè quedi net com a valor monetari.
df['Salari mensual'] = df['Salari mensual'].fillna(df.groupby('Càrrec')['Salari mensual'].transform('mean'))
df['Salari mensual'] = df['Salari mensual'].round(2)

print(f"S'han imputat {nuls_abans_salari} salaris faltants (mitjana per Càrrec)")
print(df['Salari mensual'].isna().sum())  # hauria de sortir 0
```

6. **Aplicació dels límits de la política**: es limita l'increment total al màxim permès (15%, segons `max_total_increment_percent`), i es calcula finalment el nou salari ajustat per a cada empleat.

- **Resultat:** totes les files disposen ara d'informació de cost de vida i transport (0 valors nuls restants), llestes per calcular l'increment salarial.

```python
# Dividim Cost_vida_index i Transport_public_score en 3 trams (low/medium/high)
# utilitzant tercils, ja que la política salarial no especifica els llindars exactes.
df['cost_vida_categoria'] = pd.qcut(df['Cost_vida_index'], q=3, labels=['low', 'medium', 'high'])
df['transport_categoria'] = pd.qcut(df['Transport_public_score'], q=3, labels=['low', 'medium', 'high'])

# Calculem la franja d'edat de cada empleat (any actual - any de naixement)
any_actual = 2026
edat = any_actual - df['Any de Naixement']

def franja_edat(edat):
    if edat < 30:
        return '18-29'
    elif edat < 45:
        return '30-44'
    elif edat < 60:
        return '45-59'
    else:
        return '60+'

df['franja_edat'] = edat.apply(franja_edat)

# Determinem si té fills o no (basat en la columna unificada Nombre_fills)
df['te_fills'] = df['Nombre_fills'] > 0

df[['cost_vida_categoria', 'transport_categoria', 'franja_edat', 'te_fills']].head()
```

```python
# Calculem l'increment salarial total combinant els 4 factors del JSON:
# grup professional, situació familiar, franja d'edat, i cost de vida + transport.

def calcular_increment(fila):
    # 1. Increment base segons Grup Professional
    increment_grup = politica_salarial['base_increment_by_group'].get(fila['Grup Professional'], 0)
    
    # 2. Ajust segons si té fills
    clau_fills = 'has_children' if fila['te_fills'] else 'no_children'
    increment_fills = politica_salarial['children_adjustment'][clau_fills]
    
    # 3. Ajust segons franja d'edat
    increment_edat = politica_salarial['age_adjustment'][fila['franja_edat']]
    
    # 4. Ajust segons cost de vida i transport de la ciutat
    increment_cost_vida = politica_salarial['cost_of_living_adjustment'][fila['cost_vida_categoria']]
    increment_transport = politica_salarial['transport_adjustment'][fila['transport_categoria']]
    
    total = increment_grup + increment_fills + increment_edat + increment_cost_vida + increment_transport
    
    # Apliquem el límit màxim definit a la política
    maxim = politica_salarial['rules']['max_total_increment_percent']
    return min(total, maxim)

df['increment_percent'] = df.apply(calcular_increment, axis=1)

print(df['increment_percent'].describe())
```

```python
# Calculem el salari ajustat aplicant l'increment percentual calculat
df['salari_ajustat'] = (df['Salari mensual'] * (1 + df['increment_percent'] / 100)).round(2)

df[['Salari mensual', 'increment_percent', 'salari_ajustat']].head()
```

### Conclusions

- **Solució aplicada:** es calcula l'increment salarial total combinant quatre factors definits a la política salarial (`politica_salarial.json`): l'increment base segons Grup Professional, l'ajust segons si l'empleat té fills, l'ajust segons la seva franja d'edat, i l'ajust segons el cost de vida i el transport públic de la seva ciutat (dividits en trams low/medium/high mitjançant tercils, ja que la política no especificava els llindars exactes). El resultat es limita al màxim del 15% establert a les regles de la política.
- **Resultat:** l'increment mitjà obtingut és del 6,98% (rang entre 2,5% i 12%, sense arribar mai al límit màxim), i s'aplica sobre el `Salari mensual` per obtenir el `salari_ajustat` final de cada empleat.

### Creació de variables derivades

En aquest apartat es generen tres noves variables a partir de la informació ja disponible al dataset, amb l'objectiu d'obtenir indicadors més informatius que les dades brutes per si soles:

1. **Índex de dependència familiar** (`Salari mensual / (Nombre_fills + 1)`): mesura quants euros de salari corresponen a cada membre de la unitat familiar (l'empleat més els seus fills). Se suma 1 al nombre de fills per evitar dividir per zero quan l'empleat no en té, i perquè l'empleat mateix també forma part de la unitat familiar que ha de mantenir. Un valor baix indica que el salari s'ha de repartir entre més persones; un valor alt indica més marge econòmic per persona.
2. **Ràtio salari / cost de vida** (`salari_ajustat / Cost_vida_index`): el salari per si sol no permet comparar de manera justa el poder adquisitiu real entre persones que viuen en ciutats diferents, ja que un mateix import val molt més en una ciutat barata que en una de cara. Aquesta ràtio aproxima el poder adquisitiu relatiu de cada empleat, tenint en compte l'entorn on viu.
3. **Indicador d'etapa vital** (categòric: 0, 1, 2): resumeix en una única variable la combinació d'edat i càrregues familiars d'un empleat, classificant-lo en:
   - **0 - Sense càrregues:** sense fills i en una franja d'edat jove (18-29 anys).
   - **1 - Etapa activa:** amb responsabilitats moderades (té fills i/o es troba en una franja d'edat intermèdia, 30-59 anys).
   - **2 - Alta dependència:** combinació de factors que suggereixen més càrrega econòmica/familiar (per exemple, tenir un nombre elevat de fills, independentment de l'edat).

Aquestes variables permeten anàlisis posteriors més matisades que les dades originals per separat, com ara identificar perfils d'empleats amb menys marge econòmic relatiu o comparar el poder adquisitiu real entre ciutats/països.

```python
# 1. Índex de dependència familiar: salari repartit entre l'empleat i els seus fills.
# Sumem 1 al nombre de fills perquè l'empleat mateix també forma part de la unitat
# familiar, i per evitar dividir per zero quan no té fills.
df['index_dependencia_familiar'] = (df['Salari mensual'] / (df['Nombre_fills'] + 1)).round(2)

# 2. Ràtio salari / cost de vida: aproxima el poder adquisitiu relatiu de l'empleat
# tenint en compte l'entorn on viu (una ciutat més cara "consumeix" més el salari).
df['ratio_salari_cost_vida'] = (df['salari_ajustat'] / df['Cost_vida_index']).round(2)

# 3. Indicador d'etapa vital: combina franja d'edat i nombre de fills en una
# única categoria (0=sense càrregues, 1=etapa activa, 2=alta dependència).
def etapa_vital(fila):
    if fila['Nombre_fills'] >= 3:
        return 2  # alta dependència: nombre elevat de fills
    elif fila['franja_edat'] == '18-29' and fila['Nombre_fills'] == 0:
        return 0  # sense càrregues: jove i sense fills
    else:
        return 1  # etapa activa: la resta de casos

df['etapa_vital'] = df.apply(etapa_vital, axis=1)

print(df[['index_dependencia_familiar', 'ratio_salari_cost_vida', 'etapa_vital']].describe())
print(df['etapa_vital'].value_counts())
```

```python
# Cas límit 1: persones sense Nombre_fills (per tant, sense índex de dependència calculable)
print("Files amb Nombre_fills nul:", df['Nombre_fills'].isna().sum())

# Cas límit 2: valor màxim de la ràtio salari/cost de vida
print(df.nlargest(3, 'ratio_salari_cost_vida')[['Salari mensual', 'Cost_vida_index', 'ratio_salari_cost_vida']])
```

### Conclusions — Detecció de casos límit i incoherències en les variables derivades

- **`index_dependencia_familiar`:** presenta 28 valors nuls, corresponents a empleats sense cap informació disponible sobre el nombre de fills (ni `Fills` ni `No Fills` tenien valor a l'origen, i per tant `Nombre_fills` va quedar buit després de la unificació de columnes). Aquests casos no es poden calcular ni s'han d'imputar arbitràriament, ja que assumir un nombre de fills sense cap indici seria inventar informació.
- **`ratio_salari_cost_vida`:** els valors més alts (fins a 65,4) corresponen a empleats amb salaris molt elevats (3.400-3.700 €) que resideixen en ciutats amb un cost de vida baix (60-65 punts). No es tracta d'errors, sinó de casos reals i coherents: un salari alt en un entorn econòmicament assequible es tradueix lògicament en un poder adquisitiu relatiu molt superior a la mitjana.
- **`etapa_vital`:** la distribució (90 en etapa 0, 682 en etapa 1, 133 en etapa 2) sembla raonable i sense valors inesperats, ja que la majoria d'empleats es troben en l'etapa activa intermèdia.

```python
# Pas 1: Calculem l'edat numèrica de cada empleat (necessària per a l'edat mediana del resum)
any_actual = 2026
df['Edat'] = any_actual - df['Any de Naixement']
```

```python
# Pas 3: Creem un DataFrame resum amb estadístiques clau per Grup Professional:
# nombre de treballadors, salari mitjà i edat mediana.
resum_grups = df.groupby('Grup Professional').agg(
    nombre_treballadors=('DNI', 'count'),
    salari_mitja=('Salari mensual', 'mean'),
    edat_mediana=('Edat', 'median')
).round(2)

resum_grups.to_csv('resum_grups_professionals.csv')
resum_grups
```

```python
# (EXTRA) Consolidem totes les taules en un únic fitxer Excel,
# amb una pestanya per a cada Grup Professional i una pestanya addicional
# amb el resum agregat, per facilitar la consulta conjunta.

with pd.ExcelWriter('dades_per_grups.xlsx', engine='openpyxl') as writer:
    for grup in df['Grup Professional'].unique():
        df_grup = df[df['Grup Professional'] == grup]
        nom_pestanya = grup.replace(' ', '_')  # els noms de pestanya no poden tenir espais problemàtics
        df_grup.to_excel(writer, sheet_name=nom_pestanya, index=False)

    resum_grups.to_excel(writer, sheet_name='Resum')

print("Fitxer 'dades_per_grups.xlsx' creat amb totes les pestanyes")
```
